```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter, minimum_filter, maximum_filter

def apply_gaussian_filter(image, sigma, border_type=cv2.BORDER_DEFAULT):
    return cv2.GaussianBlur(image, (0, 0), sigma, borderType=border_type)

def apply_box_filter(image, kernel_size):
    return cv2.blur(image, (kernel_size, kernel_size))

def apply_mean_filter(image, kernel_size):
    return cv2.boxFilter(image, -1, (kernel_size, kernel_size))

def apply_laplacian_filter(image):
    return cv2.Laplacian(image, cv2.CV_64F, ksize=3)

def apply_minimum_filter(image, kernel_size):
    return minimum_filter(image, size=(kernel_size, kernel_size))

def apply_maximum_filter(image, kernel_size):
    return maximum_filter(image, size=(kernel_size, kernel_size))

def apply_median_filter(image, kernel_size):
    return cv2.medianBlur(image, kernel_size)

def apply_midpoint_filter(image, kernel_size):
    min_filtered = minimum_filter(image, size=(kernel_size, kernel_size))
    max_filtered = maximum_filter(image, size=(kernel_size, kernel_size))
    return ((min_filtered.astype(np.float32) + max_filtered.astype(np.float32)) / 2).astype(np.uint8)

def plot_images(original, filtered, title1, title2):
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(original, cmap='gray')
    plt.title(title1)
    plt.axis('off')
    plt.subplot(1, 2, 2)
    plt.imshow(filtered, cmap='gray')
    plt.title(title2)
    plt.axis('off')
    plt.show()


```

```python
# Load grayscale image
image = cv2.imread('/mnt/d/MSAIM/trimister-3 msaiml/Computer_vision/Computer_vision_practical/ImageEnhancement/image.png', cv2.IMREAD_GRAYSCALE)

# Ensure image is loaded
if image is None:
    raise ValueError("Image not found. Please check the file path.")

# Example usage
sigma_values = [1, 3, 5]
for sigma in sigma_values:
    filtered = apply_gaussian_filter(image, sigma, cv2.BORDER_REFLECT)
    plot_images(image, filtered, 'Original', f'Gaussian Filter (σ={sigma})')

laplacian = apply_laplacian_filter(image)
plot_images(image, laplacian, 'Original', 'Laplacian Filter')

kernel_sizes = [3, 5, 7]
for k in kernel_sizes:
    min_filtered = apply_minimum_filter(image, k)
    plot_images(image, min_filtered, 'Original', f'Min Filter {k}x{k}')

    max_filtered = apply_maximum_filter(image, k)
    plot_images(image, max_filtered, 'Original', f'Max Filter {k}x{k}')

    median_filtered = apply_median_filter(image, k)
    plot_images(image, median_filtered, 'Original', f'Median Filter {k}x{k}')

    midpoint_filtered = apply_midpoint_filter(image, k)
    plot_images(image, midpoint_filtered, 'Original', f'Midpoint Filter {k}x{k}')
```
