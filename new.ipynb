```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the old image
image = cv2.imread("image.png")

# Check if the image was loaded successfully
if image is None:
    raise FileNotFoundError("The image file was not found. Please check the path.")

# Convert from BGR to RGB for matplotlib display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Denoise the image while preserving color
# Parameters: (src, dst, h, hForColor, templateWindowSize, searchWindowSize)
denoised = cv2.fastNlMeansDenoisingColored(image, None, 10, 10, 7, 21)

# Sharpening filter
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])
sharpened = cv2.filter2D(denoised, -1, kernel)

# Convert denoised image to RGB for matplotlib display
sharpened_rgb = cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB)

# Display original and restored images using matplotlib
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(image_rgb)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(sharpened_rgb)
ax[1].set_title("Restored Image")
ax[1].axis("off")

plt.show()

# Save the restored image
cv2.imwrite("restored_photo.jpg", sharpened)

```
